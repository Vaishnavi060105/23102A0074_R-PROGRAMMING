**Custom Median-Imputation Function**

```r
median_impute <- function(x) {

  if (!is.numeric(x)) {
    stop("Input must be numeric.")
  }

  median_value <- median(x, na.rm = TRUE)

  x[is.na(x)] <- median_value

  return(x)
}
```

```r
numeric_columns <- c(
  "age",
  "fnlwgt",
  "education_num",
  "capital_gain",
  "capital_loss",
  "hours_per_week"
)

for (col in numeric_columns) {
  data[[col]] <- median_impute(data[[col]])
}
```
