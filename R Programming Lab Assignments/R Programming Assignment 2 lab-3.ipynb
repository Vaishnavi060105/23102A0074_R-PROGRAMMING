**Creating BP Cleaning Function**

```r
clean_bp <- function(bp) {

  if (is.na(bp)) {
    return(NA)

  } else if (bp < 0) {
    return(NA)

  } else if (bp > 250) {
    return(250)

  } else {
    return(bp)
  }
}
```

**Testing the function**

```r
test_values <- c(-100, NA, 120, 200, 300, 350)

sapply(test_values, clean_bp)
```

**Applying the cleaning function**

```r
data$trestbps_clean <- sapply(
  data$trestbps,
  clean_bp
)

head(data[, c("trestbps", "trestbps_clean")], 10)
```

**Error Handling using tryCatch()**

```r
safe_mean_bp <- function(bp) {

  tryCatch({

    if (all(is.na(bp))) {
      stop("All BP values are missing.")
    }

    mean(bp, na.rm = TRUE)

  }, warning = function(w) {
    message("Warning: ", w$message)
    NA
  }, error = function(e) {
    message("Error: ", e$message)
    NA
  })
}
```

```r
mean_bp <- safe_mean_bp(data$trestbps_clean)

cat("Mean cleaned BP:", mean_bp, "\n")
```

**Cholesterol / BP Ratio**

```r
safe_ratio <- function(chol, bp) {

  tryCatch({

    if (is.na(chol) || is.na(bp)) {
      stop("Cholesterol or BP is missing.")
    }

    if (bp <= 0) {
      stop("BP denominator must be greater than zero.")
    }

    chol / bp

  }, error = function(e) {
    message("Invalid ratio: ", e$message)
    NA
  })
}
```

```r
ratio1 <- safe_ratio(200, 120)
ratio2 <- safe_ratio(250, 0)
ratio3 <- safe_ratio(220, NA)
ratio4 <- safe_ratio(200, -50)

cat("Ratio 1:", ratio1, "\n")
cat("Ratio 2:", ratio2, "\n")
cat("Ratio 3:", ratio3, "\n")
cat("Ratio 4:", ratio4, "\n")
```

**Loop-Based Cleaning**

```r
bp_loop <- data$trestbps

start_loop <- Sys.time()

for (i in 1:length(bp_loop)) {

  if (is.na(bp_loop[i])) {
    next

  } else if (bp_loop[i] < 0) {
    bp_loop[i] <- NA

  } else if (bp_loop[i] > 250) {
    bp_loop[i] <- 250
  }
}

end_loop <- Sys.time()

loop_time <- end_loop - start_loop

bp_loop[1:10]
```

**Vectorized Cleaning**

```r
bp_vectorized <- data$trestbps

start_vector <- Sys.time()

bp_vectorized[bp_vectorized < 0] <- NA

bp_vectorized[bp_vectorized > 250] <- 250

end_vector <- Sys.time()

vector_time <- end_vector - start_vector

bp_vectorized[1:10]
```

**Comparing Execution Time**

```r
comparison <- data.frame(
  Method = c("For Loop", "Vectorized"),
  Execution_Time = c(
    as.numeric(loop_time),
    as.numeric(vector_time)
  )
)

comparison
```

**Validating Cleaned Data**

```r
cleaned_bp <- bp_vectorized

missing_count <- sum(is.na(cleaned_bp))

minimum_bp <- min(cleaned_bp, na.rm = TRUE)

maximum_bp <- max(cleaned_bp, na.rm = TRUE)

mean_bp <- mean(cleaned_bp, na.rm = TRUE)

median_bp <- median(cleaned_bp, na.rm = TRUE)

cat("Missing BP values:", missing_count, "\n")
cat("Minimum BP:", minimum_bp, "\n")
cat("Maximum BP:", maximum_bp, "\n")
cat("Mean BP:", mean_bp, "\n")
cat("Median BP:", median_bp, "\n")
```

**Verifying No Invalid Values Remain**

```r
negative_values <- sum(cleaned_bp < 0, na.rm = TRUE)

values_above_250 <- sum(cleaned_bp > 250, na.rm = TRUE)

cat("Negative BP values:", negative_values, "\n")
cat("BP values greater than 250:", values_above_250, "\n")
```
