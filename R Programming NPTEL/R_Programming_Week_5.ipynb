Example : How to work with missing data

```r
x = c(11,NA,13, NA)
mean(x)
mean (x, na.rm = TRUE)
```

Missing data : Location of missing values

```r
x = c(11,NA,13,NA)
x
which(is.na(x))
```

```r
x = c(11,NA,13,NA)
sum(is.na(x))
```

Missing data : Finding complete cases

```r
x = c(11,NA,13,NA)
x
complete.cases(x)
```

Missing data : finding complete data set

```r
x = c(11,NA,13,NA)
y = na.omit(x)
y
```

Missing Data : Handling values

```r
x = c(11,NA,13,NA)
xy = na.omit(x)
y
mean(x)
mean(y)
```

Conditional execution:if else()

```r
x = 5
if (x ==3) {x = x -1} else {x = 2*x}

x
```

```r
x = 3
if ( x==3 ) {x = x-1} else {x = 2*x}
x
```

```r
x = 6
if(x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
x = 2
if(x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

Conditional execution: Nested if else if()

```r
x = 5
if (x ==3) {
  x = x-1
} else if (x < 3) {
  x = x+5
} else {x = 2*x}
x
```

```r
x = 2
if (x ==3) {
  x = x-1
} else if (x < 3) {
  x = x+5
}else { x = 2*x }
x
```

```r
x = 3
if (x == 3) {
  x = x-1
} else if (x < 3) {
  x = x+5
} else { x = 2 *x}
x
```
